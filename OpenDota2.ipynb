```python
def realiza_request(url, method='get'):
  try:
    match method.lower():
      case 'get':
        response = requests.get(url)
      case 'post':
        response = requests.post(url)

    # Verifica se a requisição foi bem sucedida (Status 200)
    response.raise_for_status()
    json_response = response.json()

    if not json_response:
        print("Sem resposta")
        return 404
    return json_response
  except requests.exceptions.RequestException as e:
    if e.response.status_code == 404:
      print(f"Erro ao acessar a API: {e}")
      return 404
    if e.response.status_code in [429, 500]:
      print(f"Erro ao acessar a API: {e}")
      time.sleep(65)
      return realiza_request(url)
    else:
      print(f"Erro ao acessar a API: {e}")
      return None
```

```python
def player_info(account_id):
  return realiza_request(f"https://api.opendota.com/api/players/{account_id}")
```

```python
def get_top_players_info(ids):
  ids_lidos = []
  ids_com_erro = []
  indice=0
  try:
    for id in ids:

      info = player_info(id)

      if info == None:
        raise Exception()
      if info == 404:
        ids_com_erro.append(id)
        continue
      
      with open(f'info_players/id{id}.json', 'w', encoding='utf-8') as arquivo:
        json.dump(info, arquivo, ensure_ascii=False, indent=2)
      indice+=1
      ids_lidos.append(int(id))

  except Exception as e:
    print(f"erro no id {id} de indice {indice}")
    print(e)
  finally:
    return ids_lidos, ids_com_erro
```

```python

def buscar_chat_partida(match_id):
    url = f"https://api.opendota.com/api/matches/{match_id}"

    try:
        print(f"Buscando detalhes da partida {match_id}...")
        response = requests.get(url)
        response.raise_for_status()

        dados_partida = response.json()
        chat = dados_partida.get('chat')

        if chat is None:
            return "O chat não está disponível para esta partida (pode não ter sido parseada)."

        if len(chat) == 0:
            return "Nenhuma mensagem de chat encontrada nesta partida."

        # Formatando as mensagens para leitura
        mensagens_formatadas = []
        for msg in chat:
            # 'type' pode ser 'chat' (texto normal) ou 'chatwheel' (roda de mensagens)
            tipo = msg.get('type')
            tempo = msg.get('time') # Tempo em segundos desde o início do jogo
            jogador = msg.get('unit', f"Jogador {msg.get('player_slot')}")
            texto = msg.get('key', '') #if tipo == 'chatwheel' else msg.get('key', '')

            # Converter tempo em formato MM:SS
            minutos = int(tempo // 60)
            segundos = int(tempo % 60)
            timestamp = f"{minutos:02d}:{segundos:02d}"

            mensagens_formatadas.append(f"[{timestamp}] {jogador}: {texto}")

        return mensagens_formatadas

    except requests.exceptions.RequestException as e:
        return f"Erro ao acessar a API: {e}"

# Exemplo de uso:
ID_PARTIDA = "8638178478" # Substitua por um ID real
historico_chat = buscar_chat_partida(ID_PARTIDA)

if isinstance(historico_chat, list):
    for linha in historico_chat:
        print(linha)
else:
    print(historico_chat)
```

```python
def integrar_chats_ao_dataframe(df_partidas):
    if df_partidas is None or df_partidas.empty:
        print("DataFrame vazio ou inválido.")
        return None

    print(f"Iniciando a busca de chat para {len(df_partidas)} partidas...")

    # Criamos uma lista para armazenar os resultados temporariamente
    todos_os_chats = []

    for index, row in df_partidas.iterrows():
        match_id = row['match_id']

        # Chama a função que criamos anteriormente
        resultado_chat = buscar_chat_partida(match_id)

        # Se o resultado for uma lista (mensagens), juntamos em uma única string
        # Se for erro ou string de aviso, salvamos a mensagem de erro
        if isinstance(resultado_chat, list):
            chat_texto = " | ".join(resultado_chat)
        else:
            chat_texto = resultado_chat

        todos_os_chats.append(chat_texto)

        # Pausa de 1 segundo entre requisições para respeitar o limite da API gratuita
        time.sleep(1)
        print(f"Processado {index + 1}/{len(df_partidas)} (Match ID: {match_id})")

    # Adiciona a nova coluna ao DataFrame
    df_partidas['chat_log'] = todos_os_chats

    return df_partidas

# --- EXECUÇÃO ---

# 1. Recupera as últimas 10 partidas de um jogador
df_inicial = buscar_partidas_jogador(PLAYER_ID)

# 2. Enriquece o DataFrame com o chat
df_final = integrar_chats_ao_dataframe(df_inicial)

# 3. Exibe o resultado
print("\nDataFrame Atualizado:")
print(df_final[['match_id', 'chat_log']].head())
```

```python
def solicitar_parse_partida(match_id):
    # O endpoint para POST é /requests/{match_id}
    url = f"https://api.opendota.com/api/requests/{match_id}"


    try:
        print(f"Enviando solicitação de parse para a partida {match_id}...")

        # Note que usamos requests.post() em vez de get()
        response = requests.post(url)

        # Verifica se a API aceitou a requisição (Status 200 ou 201)
        response.raise_for_status()

        dados = response.json()

        # A API retorna um objeto confirmando que a partida entrou na fila
        if 'job' in dados:
            job_id = dados['job'].get('jobId')
            print(f"Sucesso! Partida colocada na fila de processamento.")
            print(f"ID do Trabalho (Job ID): {job_id}")
            return dados
        else:
            return "A resposta foi recebida, mas o formato foi inesperado."

    except requests.exceptions.RequestException as e:
        # Erro 429 geralmente significa que você atingiu o limite de parses gratuitos
        return f"Erro ao solicitar parse: {e}"

# Exemplo de uso
resultado = solicitar_parse_partida(match_id)
print(resultado)
```

```python
def consolidar_json_players(caminho_pasta="info_players"):
    # Lista para armazenar os dados de cada jogador
    lista_dados = []
    
    # Verifica se a pasta existe
    if not os.path.exists(caminho_pasta):
        print(f"Erro: A pasta '{caminho_pasta}' não foi encontrada.")
        return None

    # Lista todos os arquivos dentro da pasta
    arquivos = os.listdir(caminho_pasta)
    
    # Filtra apenas os arquivos que terminam com .json
    arquivos_json = [f for f in arquivos if f.endswith('.json')]
    
    print(f"Encontrados {len(arquivos_json)} arquivos JSON na pasta '{caminho_pasta}'.")

    # Loop para ler arquivo por arquivo
    for nome_arquivo in arquivos_json:
        caminho_completo = os.path.join(caminho_pasta, nome_arquivo)
        
        try:
            with open(caminho_completo, 'r', encoding='utf-8') as arquivo:
                dados_jogador = json.load(arquivo)
                
                # Se o JSON for um dicionário único (um jogador por arquivo)
                if isinstance(dados_jogador, dict):
                    lista_dados.append(dados_jogador)
                # Se o JSON já for uma lista de dicionários dentro do arquivo
                elif isinstance(dados_jogador, list):
                    lista_dados.extend(dados_jogador)
                    
        except Exception as e:
            print(f"Erro ao ler o arquivo {nome_arquivo}: {e}")

    # Cria o DataFrame a partir da lista consolidada
    if lista_dados:
        df = pd.DataFrame(lista_dados)
        print("DataFrame gerado com sucesso!")
        return df
    else:
        print("Nenhum dado válido foi extraído.")
        return pd.DataFrame()
```

```python

df_players = consolidar_json_players()

df_players
```

```python
df_profile_expandido = pd.json_normalize(df_players['profile'])

# 3. Junta o novo DataFrame expandido com o original e remove a coluna de dicionários original
df_players = pd.concat([df_profile_expandido, df_players.drop(columns=['profile'])], axis=1)

# Visualizando o resultado
df_players
```
